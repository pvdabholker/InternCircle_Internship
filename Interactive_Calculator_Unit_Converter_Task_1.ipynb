# Task 1 - Interactive Calculator & Unit Converter

## InternCircle Data Science & AI Internship

This project is a beginner-friendly command-line program developed
using Python.

### Features

- Basic arithmetic operations
- Currency conversion
- Kilometres to miles conversion
- Celsius to Fahrenheit conversion
- Input validation using while loops
- Menu-driven user interaction
- Functions for different operations

### Technologies Used

- Python
- Google Colab

```python
def arithmetic_calculator():
    print("\n--- Arithmetic Calculator ---")

    while True:
        try:
            num1 = float(input("Enter first number: "))
            num2 = float(input("Enter second number: "))
            break
        except ValueError:
            print("Invalid input. Please enter numbers only.")

    print("\nChoose an operation:")
    print("1. Addition (+)")
    print("2. Subtraction (-)")
    print("3. Multiplication (*)")
    print("4. Division (/)")

    while True:
        choice = input("Enter your choice (1-4): ")

        if choice in ["1", "2", "3", "4"]:
            break

        print("Invalid choice. Please enter a number from 1 to 4.")

    if choice == "1":
        result = num1 + num2
        print("Result:", result)

    elif choice == "2":
        result = num1 - num2
        print("Result:", result)

    elif choice == "3":
        result = num1 * num2
        print("Result:", result)

    elif choice == "4":
        if num2 == 0:
            print("Error: Cannot divide by zero.")
        else:
            result = num1 / num2
            print("Result:", result)
```

```python
def currency_converter():
    print("\n--- Currency Converter ---")

    print("Available conversions:")
    print("1. INR to USD")
    print("2. USD to INR")
    print("3. INR to EUR")
    print("4. EUR to INR")

    while True:
        choice = input("Enter your choice (1-4): ")

        if choice in ["1", "2", "3", "4"]:
            break

        print("Invalid choice. Please select 1-4.")

    while True:
        try:
            amount = float(input("Enter amount: "))

            if amount < 0:
                print("Amount cannot be negative.")
                continue

            break

        except ValueError:
            print("Invalid input. Please enter a number.")

    # Example fixed conversion rates
    # Rates can be updated when required.
    INR_TO_USD = 0.012
    USD_TO_INR = 83.50
    INR_TO_EUR = 0.011
    EUR_TO_INR = 90.50

    if choice == "1":
        result = amount * INR_TO_USD
        print(f"{amount:.2f} INR = {result:.2f} USD")

    elif choice == "2":
        result = amount * USD_TO_INR
        print(f"{amount:.2f} USD = {result:.2f} INR")

    elif choice == "3":
        result = amount * INR_TO_EUR
        print(f"{amount:.2f} INR = {result:.2f} EUR")

    elif choice == "4":
        result = amount * EUR_TO_INR
        print(f"{amount:.2f} EUR = {result:.2f} INR")
```

```python
def km_to_miles():
    print("\n--- Kilometres to Miles Converter ---")

    while True:
        try:
            km = float(input("Enter distance in kilometres: "))

            if km < 0:
                print("Distance cannot be negative.")
                continue

            break

        except ValueError:
            print("Invalid input. Please enter a number.")

    miles = km * 0.621371

    print(f"{km:.2f} km = {miles:.2f} miles")
```

```python
def celsius_to_fahrenheit():
    print("\n--- Celsius to Fahrenheit Converter ---")

    while True:
        try:
            celsius = float(input("Enter temperature in Celsius: "))
            break

        except ValueError:
            print("Invalid input. Please enter a number.")

    fahrenheit = (celsius * 9 / 5) + 32

    print(f"{celsius:.2f}°C = {fahrenheit:.2f}°F")
```

```python
def main():
    while True:
        print("\n===================================")
        print(" Interactive Calculator & Converter")
        print("===================================")

        print("1. Arithmetic Calculator")
        print("2. Currency Converter")
        print("3. Kilometres to Miles")
        print("4. Celsius to Fahrenheit")
        print("5. Exit")

        choice = input("\nEnter your choice (1-5): ")

        if choice == "1":
            arithmetic_calculator()

        elif choice == "2":
            currency_converter()

        elif choice == "3":
            km_to_miles()

        elif choice == "4":
            celsius_to_fahrenheit()

        elif choice == "5":
            print("\nThank you for using the calculator!")
            break

        else:
            print("Invalid choice. Please select a number from 1 to 5.")
```

```python
main()
```
